```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("hydroponics_datasets/IoTData --Raw--.csv")

df['timestamp'] = pd.to_datetime(df['timestamp'])

print(df.info())
print("\nStatistik Dasar Sensor:")
print(df.describe())
```

```python
actuator_cols = ['pH_reducer', 'add_water', 'nutrients_adder', 'humidifier', 'ex_fan']

plt.figure(figsize=(15, 8))
for i, col in enumerate(actuator_cols, 1):
    plt.subplot(2, 3, i)
    
    counts = df[col].value_counts(normalize=True) * 100

    sns.countplot(x=col, data=df, palette='Set2')
    plt.title(f'{col}\nON: {counts.get("ON", 0):.1f}% | OFF: {counts.get("OFF", 0):.1f}%')
    plt.ylabel('Jumlah Kejadian')

plt.tight_layout()
plt.show()
```
